```python
import os

import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv("../.env")
engine = create_engine(
    f"postgresql+psycopg://{os.environ['POSTGRES_USER']}:{os.environ['POSTGRES_PASSWORD']}"
    f"@{os.getenv('POSTGRES_HOST', 'localhost')}:{os.getenv('POSTGRES_PORT', '5432')}/{os.environ['POSTGRES_DB']}"
)
```

```python
pd.read_sql("SELECT eval_set, count(*) AS orders, count(DISTINCT user_id) AS users FROM raw.orders GROUP BY eval_set", engine)
```

```python
pd.read_sql("""
SELECT p.product_name, p.department, p.aisle, count(*) AS times_ordered
FROM raw.order_products_prior op
JOIN raw.v_products p USING (product_id)
GROUP BY 1, 2, 3
ORDER BY times_ordered DESC
LIMIT 10
""", engine)
```
